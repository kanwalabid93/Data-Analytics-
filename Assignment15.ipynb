## List all cities where the area is greater than 5000, along with their population.

```python
q = '''select cities.name, cities.population
from cities
join facts on cities.facts_id = facts.id
WHERE facts.area > 5000'''
sql_query(q)
```

## 2. List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%

```python
q = '''
select cities.name
from cities
join facts on cities.facts_id = facts.id
where cities.capital = 1
and facts.death_rate > 2
and facts.birth_rate < 20
'''
sql_query(q)
```

## 3. List the cities with a population growth rate between 1% and 5%, sorted in desc by population growth.

```python
q = '''
select cities.name, facts.population_growth
from cities
join facts on cities.facts_id = facts.id
where facts.population_growth between 1 and 5
order by facts.population_growth DESC
'''
sql_query(q)
```

## 4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities table. The query should return the country name and city name. Using left join

```python
q = '''
select facts.name as country_name, cities.name as city_name
from facts
left join cities on facts.id = cities.facts_id
where facts.name Like 'A%'
and cities.id is null
'''

sql_query(q)
```

## 5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
q = '''
select cities.name, facts.death_rate
from cities
left join facts on cities.facts_id = facts.id
where cities.capital = 1
'''
sql_query(q)
```
