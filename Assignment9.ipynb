```python
import csv
```

```python
open_file=open ('student data.csv',mode='r')
```

```python
reader=csv.reader(open_file)
```

```python
data=list(reader)
```

```python
data[0].append('Percentage')
data[0].append('Grades')
```

```python
student_data=data[1:]
```

```python
for per in student_data:
    chem=int(per[2])
    phy=int(per[3])
    math=int(per[4])
    percentage=round((chem+phy+math)/300*100,2)
    per.append(percentage)

    if percentage >= 80:
        grade = 'A'
    elif percentage >= 70:
        grade = 'B'
    elif percentage >= 60:
        grade = 'C'
    elif percentage >= 50:
        grade = 'D'
    else:
        grade = 'F'

    per.append(grade)
```

```python
open_write=open('data.csv', mode='w', newline='')
```

```python
writer=csv.writer(open_write)
writer.writerows(data)
open_write.close()
```
