```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience
    def describe_employee(self):
        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Employee Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Years of Experience:", self.year_of_experience)
    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to {self.year_of_experience} years.")
    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s new salary is {self.salary}.")

employee101 = Employee( "Ali Khan",101,"Lahore",50000,"Software Engineer",2020,"BS Computer Science",4)
print("Name:", employee101.name)
print("Employee Number:", employee101.number)
print("Salary:", employee101.salary)
print("Post:", employee101.post)

employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)

employees = [
    Employee("Ahmed", 102, "Karachi", 45000, "Developer", 2019, "BS IT", 5),
    Employee("Sara", 103, "Islamabad", 55000, "Manager", 2018, "MBA", 7),
    Employee("Usman", 104, "Peshawar", 40000, "Designer", 2021, "BS Design", 3),
    Employee("Ayesha", 105, "Quetta", 60000, "Team Lead", 2017, "MS CS", 8),
    Employee("Bilal", 106, "Multan", 48000, "Tester", 2020, "BS SE", 4),
    Employee("Hina", 107, "Faisalabad", 52000, "HR Officer", 2019, "MBA HR", 5),
    Employee("Zain", 108, "Sialkot", 47000, "Network Engineer", 2022, "BS Networks", 2),
    Employee("Fatima", 109, "Hyderabad", 65000, "Project Manager", 2016, "MS Management", 10),
    Employee("Hamza", 110, "Gujranwala", 43000, "Support Engineer", 2021, "BS CS", 3),
    Employee("Noor", 111, "Bahawalpur", 70000, "Data Scientist", 2015, "MS AI", 11)
]

print("\n===== Employee Information =====")
for emp in employees:
    emp.describe_employee()

print("\n===== Updating Employee Records =====")
employees[0].increase_salary(3000)
employees[0].update_year_of_experience(1)

employees[3].increase_salary(7000)
employees[3].update_year_of_experience(2)

employees[7].increase_salary(10000)
employees[7].update_year_of_experience(1)

print("\n===== Updated Employee Details =====")
employees[0].describe_employee()
employees[3].describe_employee()
employees[7].describe_employee()
```
